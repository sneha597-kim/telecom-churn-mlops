```python
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")
#Why did we use errors="coerce"?
#Instead of throwing an error when it finds " ", Pandas converts it to NaN.
#NaN is much easier to handle because Pandas has built-in functions for missing values.
```
